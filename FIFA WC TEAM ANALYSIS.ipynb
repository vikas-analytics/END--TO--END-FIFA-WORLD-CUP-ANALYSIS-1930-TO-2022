```python
import numpy as np

matches["Result"] = np.select(
    [
        matches["home_team_score"] > matches["away_team_score"],
        matches["home_team_score"] < matches["away_team_score"]
    ],
    [
        "Home Win",
        "Away Win"
    ],
    default="Draw"
)
```

```python
#TEAM LEVEL DATASET 
```

```python
home = pd.DataFrame({                 #HOME DATA OF TEAMS
    "Match_ID": matches["match_id"],
    "Year": matches["match_date"].dt.year,
    "Team": matches["home_team_name"],
    "Opponent": matches["away_team_name"],
    "Goals_Scored": matches["home_team_score"],
    "Goals_Conceded": matches["away_team_score"],
    "Result": matches["Result"]
})
```

```python
home["Points"] = np.select(
    [
        home["Result"] == "Home Win",
        home["Result"] == "Draw"
    ],
    [
        3,
        1
    ],
    default=0
)
```

```python
away = pd.DataFrame({                     #AWAY DATA OF TEAMS
    "Match_ID": matches["match_id"],
    "Year": matches["match_date"].dt.year,
    "Team": matches["away_team_name"],
    "Opponent": matches["home_team_name"],
    "Goals_Scored": matches["away_team_score"],
    "Goals_Conceded": matches["home_team_score"],
    "Result": matches["Result"]
})
```

```python
away["Result"] = np.select(
    [
        matches["Result"] == "Away Win",
        matches["Result"] == "Draw"
    ],
    [
        "Win",
        "Draw"
    ],
    default="Loss"
)
```

```python
home["Result"] = np.select(
    [
        matches["Result"] == "Home Win",
        matches["Result"] == "Draw"
    ],
    [
        "Win",
        "Draw"
    ],
    default="Loss"
)
```

```python
away["Points"] = np.select(
    [
        away["Result"] == "Win",
        away["Result"] == "Draw"
    ],
    [
        3,
        1
    ],
    default=0
)
```

```python
# COMBINING HOME + AWAY PERFORMANCE
```

```python
team_matches = pd.concat(
    [home, away],
    ignore_index=True
)
```

```python
team_matches.head(5)
```

```python
team_matches.shape
```

```python
# CREATING OF GOAL DIFFERENCE
```

```python
team_matches["Goal_Difference"] = (
    team_matches["Goals_Scored"] -
    team_matches["Goals_Conceded"]
)
```

```python
team_matches["Goal_Difference"]
```

```python
#Win/Draw/Loss Flags
```

```python
team_matches["Win"] = (
    team_matches["Result"] == "Win"
).astype(int)

team_matches["Draw"] = (
    team_matches["Result"] == "Draw"
).astype(int)

team_matches["Loss"] = (
    team_matches["Result"] == "Loss"
).astype(int)
```

```python
#Final Team Performance Dataset
```

```python
team_performance = (
    team_matches
    .groupby("Team")
    .agg(
        Matches_Played=("Match_ID", "count"),
        Wins=("Win", "sum"),
        Draws=("Draw", "sum"),
        Losses=("Loss", "sum"),
        Goals_Scored=("Goals_Scored", "sum"),
        Goals_Conceded=("Goals_Conceded", "sum"),
        Points=("Points", "sum")
    )
    .reset_index()
)
```

```python
team_performance["Goal_Difference"] = (
    team_performance["Goals_Scored"] -
    team_performance["Goals_Conceded"]
)
```

```python
team_performance["Win_Rate"] = (
    team_performance["Wins"] /
    team_performance["Matches_Played"] * 100
)
```

```python
team_performance["Points_Per_Match"] = (
    team_performance["Points"] /
    team_performance["Matches_Played"]
)
```

```python
top_wins = team_performance.sort_values(
    "Wins",
    ascending=False
).head(10)

top_wins
```

```python
top_wins = team_performance.sort_values("Wins", ascending=False).head(10)
top_wins = top_wins.sort_values("Wins")

mean, median = top_wins["Wins"].mean(), top_wins["Wins"].median()
colors = ["red" if x < 20 else "orange" if x < 40 else "green" if x < 60 else "red"
          for x in top_wins["Wins"]]

plt.figure(figsize=(12,6))
bars = plt.barh(top_wins["Team"], top_wins["Wins"], color=colors)

plt.axvline(mean, ls="--", label=f"Mean = {mean:.1f}")
plt.axvline(median, ls=":", label=f"Median = {median:.1f}")

for b in bars:
    plt.text(b.get_width()+.5, b.get_y()+b.get_height()/2, int(b.get_width()), va="center")

plt.title("Top 10 FIFA World Cup Teams by Wins")
plt.xlabel("Wins")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Win Rate Analysis
```

```python
top_win_rate = team_performance.sort_values(
    "Win_Rate",
    ascending=False
).head(10)

top_win_rate[
    ["Team", "Matches_Played", "Wins", "Win_Rate"]
]
```

```python
plt.figure(figsize=(12, 7))

# Sort from lowest to highest for horizontal bar chart
top_win_rate = top_win_rate.sort_values("Win_Rate")

# Assign colors based on win-rate ranges
colors = []

for value in top_win_rate["Win_Rate"]:
    if value >= 58:
        colors.append("red")
    elif value >= 51:
        colors.append("orange")
    else:
        colors.append("purple")

bars = plt.barh(
    top_win_rate["Team"],
    top_win_rate["Win_Rate"],
    color=colors
)

plt.title("Top 10 Teams by FIFA World Cup Win Rate", fontsize=16)
plt.xlabel("Win Rate (%)")
plt.ylabel("Team")

# Add percentage labels
for bar in bars:
    value = bar.get_width()
    plt.text(
        value + 0.5,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.1f}%",
        va="center",
        fontsize=10
    )

plt.xlim(0, 100)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

ax = sns.histplot(
    data=matches,
    x="Total_Goals",
    bins=range(
        matches["Total_Goals"].min(),
        matches["Total_Goals"].max() + 2
    ),
    kde=True,
    color="red",
    edgecolor="black"
)

# Mean and median
mean_goals = matches["Total_Goals"].mean()
median_goals = matches["Total_Goals"].median()

plt.axvline(
    mean_goals,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_goals:.2f}"
)

plt.axvline(
    median_goals,
    color="green",
    linestyle=":",
    linewidth=2,
    label=f"Median = {median_goals:.2f}"
)

# X-axis: every goal value
plt.xticks(
    range(
        matches["Total_Goals"].min(),
        matches["Total_Goals"].max() + 1
    )
)

plt.title(
    "Distribution of Total Goals per FIFA World Cup Match",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Total Goals per Match", fontsize=12)
plt.ylabel("Number of Matches", fontsize=12)

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
##This histogram shows the distribution of total goals scored in FIFA World Cup
matches.

- The X-axis represents the total number of goals scored in a match.
- The Y-axis represents the number of matches.
- The height of each bar indicates how frequently that goal total occurred.
- The KDE curve shows the overall shape of the goal distribution.
- The mean and median lines help identify the central tendency of match scoring.
```
